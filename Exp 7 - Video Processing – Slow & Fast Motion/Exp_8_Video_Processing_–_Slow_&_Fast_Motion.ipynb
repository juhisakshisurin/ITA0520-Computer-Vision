```python
filename = list(uploaded.keys())[0]

print("Uploaded video:", filename)
```

```python
cap = cv2.VideoCapture(filename)

fps = cap.get(cv2.CAP_PROP_FPS)
width = int(cap.get(cv2.CAP_PROP_FRAME_WIDTH))
height = int(cap.get(cv2.CAP_PROP_FRAME_HEIGHT))

print("FPS:", fps)
print("Width:", width)
print("Height:", height)

cap.release()
```

```python
cap = cv2.VideoCapture(filename)

fps = cap.get(cv2.CAP_PROP_FPS)
width = int(cap.get(cv2.CAP_PROP_FRAME_WIDTH))
height = int(cap.get(cv2.CAP_PROP_FRAME_HEIGHT))

fourcc = cv2.VideoWriter_fourcc(*'mp4v')

slow_video = cv2.VideoWriter(
    'slow_motion.mp4',
    fourcc,
    fps / 2,
    (width, height)
)

while True:
    ret, frame = cap.read()

    if not ret:
        break

    slow_video.write(frame)

cap.release()
slow_video.release()

print("Slow-motion video created")
```

```python
cap = cv2.VideoCapture(filename)

fps = cap.get(cv2.CAP_PROP_FPS)
width = int(cap.get(cv2.CAP_PROP_FRAME_WIDTH))
height = int(cap.get(cv2.CAP_PROP_FRAME_HEIGHT))

fourcc = cv2.VideoWriter_fourcc(*'mp4v')

fast_video = cv2.VideoWriter(
    'fast_motion.mp4',
    fourcc,
    fps,
    (width, height)
)

frame_count = 0

while True:
    ret, frame = cap.read()

    if not ret:
        break

    # Write every second frame
    if frame_count % 2 == 0:
        fast_video.write(frame)

    frame_count += 1

cap.release()
fast_video.release()

print("Fast-motion video created")
```

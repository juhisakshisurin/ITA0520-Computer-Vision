```python
from google.colab import files
uploaded = files.upload()
```

```python
import cv2
import matplotlib.pyplot as plt
```

```python
filename = list(uploaded.keys())[0]
image = cv2.imread(filename)
print("Image uploaded successfully:", filename)
```

```python
watermarked_image = image.copy()
```

```python
text = "SIMATS"
```

```python
cv2.putText(
    watermarked_image,
    text,
    (50, 100),
    cv2.FONT_HERSHEY_SIMPLEX,
    2,
    (255, 255, 255),
    3
)
```

```python
plt.figure(figsize=(10, 5))
```

```python
plt.subplot(1, 2, 1)
plt.imshow(cv2.cvtColor(image, cv2.COLOR_BGR2RGB))
plt.title("Original Image")
plt.axis("off")
```

```python
plt.subplot(1, 2, 2)
plt.imshow(cv2.cvtColor(watermarked_image, cv2.COLOR_BGR2RGB))
plt.title("Watermarked Image")
plt.axis("off")
```

```python
plt.show()
```

```python
overlay = image.copy()
```

```python
cv2.putText(
    overlay,
    "SIMATS",
    (50, 100),
    cv2.FONT_HERSHEY_SIMPLEX,
    2,
    (255, 255, 255),
    3
)
```

```python
watermarked_image = cv2.addWeighted(
    overlay, 0.4,
    image, 0.6,
    0
)
```

```python
plt.imshow(cv2.cvtColor(watermarked_image, cv2.COLOR_BGR2RGB))
plt.title("Transparent Watermark")
plt.axis("off")
plt.show()
```

```python
from google.colab import files

uploaded = files.upload()
```

```python
import cv2
import matplotlib.pyplot as plt
```

```python
filename = list(uploaded.keys())[0]

image = cv2.imread(filename)

print("Image uploaded successfully:", filename)
```

```python
# Resize to bigger size
bigger_image = cv2.resize(image, None, fx=2, fy=2, interpolation=cv2.INTER_LINEAR)

# Resize to smaller size
smaller_image = cv2.resize(image, None, fx=0.5, fy=0.5, interpolation=cv2.INTER_AREA)
```

```python
plt.imshow(cv2.cvtColor(image, cv2.COLOR_BGR2RGB))
plt.title("Original Image")
plt.axis("off")
plt.show()
```

```python
plt.imshow(cv2.cvtColor(bigger_image, cv2.COLOR_BGR2RGB))
plt.title("Bigger Image")
plt.axis("off")
plt.show()
```

```python
plt.imshow(cv2.cvtColor(smaller_image, cv2.COLOR_BGR2RGB))
plt.title("Smaller Image")
plt.axis("off")
plt.show()
```

```python
plt.figure(figsize=(15, 5))

plt.subplot(1, 3, 1)
plt.imshow(cv2.cvtColor(image, cv2.COLOR_BGR2RGB))
plt.title("Original Image")
plt.axis("off")

plt.subplot(1, 3, 2)
plt.imshow(cv2.cvtColor(bigger_image, cv2.COLOR_BGR2RGB))
plt.title("Bigger Image")
plt.axis("off")

plt.subplot(1, 3, 3)
plt.imshow(cv2.cvtColor(smaller_image, cv2.COLOR_BGR2RGB))
plt.title("Smaller Image")
plt.axis("off")

plt.show()

# Print actual dimensions
print("Original Image Size:", image.shape[1], "x", image.shape[0])
print("Bigger Image Size:", bigger_image.shape[1], "x", bigger_image.shape[0])
print("Smaller Image Size:", smaller_image.shape[1], "x", smaller_image.shape[0])
```
